```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# --- Member 7: Zoom & Shift Transformations ---
sample_img = cv2.imread(os.path.join(dataset_path, os.listdir(dataset_path)[0]))
sample_img = cv2.cvtColor(sample_img, cv2.COLOR_BGR2RGB)
sample_img = cv2.resize(sample_img, (150, 150))

# Width/Height Shift (Translation)
rows, cols, _ = sample_img.shape
shift_matrix = np.float32([[1, 0, 30], [0, 1, 30]]) # Shift by 30 pixels right and down
shifted_img = cv2.warpAffine(sample_img, shift_matrix, (cols, rows))

# EDA Visualization
fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(sample_img)
axes[0].set_title('Original Image')
axes[1].imshow(shifted_img)
axes[1].set_title('Shifted Image')
plt.show()
```

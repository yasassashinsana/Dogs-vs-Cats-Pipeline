```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

dataset_path = "/content/train" # Path of unzipped file
```

```python
# --- Member 2: Label Encoding (Updated for Kaggle Dogs vs Cats) ---
image_paths = []
labels = []

# Extract label from filename (cat = 0, dog = 1)
for file in os.listdir(dataset_path):
    if file.endswith('.jpg'):
        image_paths.append(os.path.join(dataset_path, file))
        if 'cat' in file:
            labels.append(0)
        elif 'dog' in file:
            labels.append(1)

# EDA Visualization
plt.bar(['Cats (0)', 'Dogs (1)'], [labels.count(0), labels.count(1)], color=['orange', 'blue'])
plt.title('Class Distribution')
plt.show()
```
